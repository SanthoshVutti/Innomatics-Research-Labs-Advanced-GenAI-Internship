# Food Delivery Hackathon Analysis

This notebook loads the final dataset and performs basic analysis.

```python
import pandas as pd

df = pd.read_csv('final_food_delivery_dataset.csv')
df.head()
```

```python
df.info()
```

```python
# Total revenue
df['total_amount'].sum()
```

```python
# Gold vs Regular orders
df['membership'].value_counts()
```

```python
# City-wise revenue
df.groupby('city')['total_amount'].sum().sort_values(ascending=False)
```

```python
# Cuisine-wise average order value
df.groupby('cuisine')['total_amount'].mean().sort_values(ascending=False)
```
